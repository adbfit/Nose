# Atlante del naso: dalla ricostruzione 3D alla fotografia di dissezione

Questo notebook esegue la fase fotografica (`genera_foto.py`) sulle mappe di controllo
prodotte da `atlas3d` e già salvate nella cartella `esempi/`.

**Prima di iniziare:** *Runtime → Cambia tipo di runtime → GPU (T4 o superiore)*.

L'anatomia (forma, strati, decorso delle arterie) è fissata dal modello 3D costruito sulla
letteratura; il generatore aggiunge solo la resa fotografica. Ogni immagine esce con una versione
`_verifica` (contorni anatomici sovrapposti) e un report di coerenza per tessuto: vanno
controllati da un anatomista prima dell'uso.
